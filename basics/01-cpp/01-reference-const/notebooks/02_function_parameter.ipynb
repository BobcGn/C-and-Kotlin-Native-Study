## 实验2：为什么函数参数会大量使用引用传递？
先写代码：


````cpp
// 本步骤：引入本实验需要的标准库和公开头文件。
#include <iostream>
````

````cpp
// 本步骤：通过代码演示“实验2：为什么函数参数会大量使用引用传递？”并观察结果。
// 值传递
void modify_by_value(int value){
    value = 100;
}
// 指针传递
void modify_by_pointer(int* value){
    *value = 200;
}
//引用传递
void modify_by_reference(int& value){
    value = 300;
}

{
    int a = 10;
    int b = 10;
    int c = 10;

    modify_by_value(a);
    modify_by_pointer(&b);
    modify_by_reference(c);

    std::cout << "a = " << a << '\n';
    std::cout << "b = " << b << '\n';
    std::cout << "c = " << c << '\n';
}
````

结果：
```
a = 10
b = 200
c = 300
```

现在回头来看这三个函数：
- 值传递
```C++
void modify_by_value(int value);
```
表示：
```
调用者
  │
  │ copy
  ▼
value
```
函数操作自己的对象

- 指针传递
```C++
void modify_by_pointer(int* value);
```
表示：
`给我一个对象地址`

- 引用传递
```C++
void modify_by_reference(int& value);
```
表示：
`给我借用一个已经存在的 int，而且我可能修改它`
调用链路非常自然

这就是 C API 和 C++ API 设计的不同之处：
- C API常见：
```C
void user_set_name(User* user, const char* name);
```

- C++ API常见：
```C++
void set_name(User& user, const std::string name);
```


这决定了后续开发 API 时的边界：
- C ABI（二进制接口）
```C
User*
const char*
```
- C++ 实现
```C++
User&
const std::string&
```
