## 实验4：引用与生命周期
这是本章节最重要的一个坑

````cpp
// 本步骤：引入本实验需要的标准库和公开头文件。
#include <iostream>
````

````cpp
// 本步骤：通过代码演示“实验4：引用与生命周期”并观察结果。
int& get_value(){
    int value = 42;
    return value;
}

{
    int& value = get_value();
    std::cout << value << std::endl;
}
````

上述代码出现了编译错误，原因很重要：
`get_value()`这个函数返回的是一个引用，而不是对象复制，于是这个引用指向了一个**生命周期已经结束的对象**
这就是**悬空引用**

### 参数设计规则
因此需要建立一个 C++ 参数设计规则：
1. 小型值
    - int
    - double
    - bool
    - enum
直接使用值传递：
```C++
void set_age(int age);
```   

2. 只读较大的对象
```C++
void process(const std::string& input);
```
表达：**只读 + 借用**

3. 需要修改调用者对象
```C++
void normalize(std::string& input);
```
表达：**可变 + 引用**（无额外的对象拷贝）

4. 参数运行“不存在”
这时候不适合使用引用，而应该使用指针
```C++
void process(User* user);
```
因为
```C++
user == nullptr;
```
是合法状态，所以指针在**可空场景下**很重要
