# 第一部分：引用和常量
## 1. 引用和指针

引用是已存在对象的别名。下面比较值、指针和引用，并观察通过引用修改对象的效果。


````c++
#include <iostream>
#include <stdio.h>
````

````c++
{
    // 普通变量直接保存值。
    int value = 10;

    // 指针保存对象的地址，可以重新指向其他对象，也可以为 nullptr。
    int* pointer = &value;

    // 引用是对象的别名，必须在定义时绑定到一个有效对象。
    int& reference = value;

    std::cout << "value      = " << value << '\n'
              << "*pointer   = " << *pointer << '\n'
              << "reference  = " << reference << "\n\n"
              << "&value     = " << &value << '\n'
              << "pointer    = " << pointer << '\n'
              << "&reference = " << &reference << "\n\n";

    // reference 是 value 的别名，所以通过 reference 赋值会修改 value。
    reference = 20;

    std::cout << "after reference = 20\n"
              << "value      = " << value << '\n'
              << "*pointer   = " << *pointer << '\n'
              << "reference  = " << reference << "\n\n";

    int a = 10;
    int b = 20;
    int& ref = a;

    std::cout << "before ref = b\n"
              << "a = " << a << '\n'
              << "b = " << b << "\n\n";

    // 这是把 b 的值赋给 ref 所引用的 a，不会让 ref 改为引用 b。
    ref = b;

    std::cout << "after ref = b\n"
              << "a = " << a << '\n'
              << "b = " << b << '\n';
}
````

从输出结果不难看出，三个地址`（&value, &reference, pointer)`都对应了同一个对象。
这揭示了引用的一个核心特征：
- 引用不是保存了对象的地址，从C++的语义看，`ref`是`value`的**别名**
因此：
```C++
ref = 20;
```
等价于：
```C++
value = 20;
```

### 和 C 的指针对比
在 C 中我们通常：

````c++
int c_val = 10;
int* c_ptr = &c_val;
*c_ptr = 20;
````

C++ 的引用：

````c++
int cpp_val = 10;
int& cpp_reference = cpp_val;
cpp_reference = 20;
````

虽然表面上区别不大，但是实际上的语义不同：
C Pointer：
```
ptr
 │
 │ contains address
 ▼
value
```
C++ Reference:
```
value
 ↑
 │
ref

ref 是 value 的另一个名字
```
因此引用没有
```C++
*ref
```
这种操作

直接使用
```C++
ref
```
就代表了对象本身

### 引用的两个重要限制：
#### **必须初始化**
- 正确示例：

````c++
int num = 10;
int& num_ref = num;
// 编译通过
````

- 错误示例：


````c++
int& error_ref;
// 编译错误
````

这是**指针**和**引用**的重大区别之一
指针可以：
```C++
int* ptr = nullptr;
```

而引用必须指向某个对象

#### **引用不能重新绑定**
继续查看最开始的实验：
```C++
int& ref = a;

std::cout << "a = " << a << '\n';
std::cout << "b = " << b << '\n';

ref = b;

std::cout << "after ref = b\n";

std::cout << "a = " << a << '\n';
std::cout << "b = " << b << '\n';
```
我们可能下意识地认为：
```C++
ref = b;
```
会将`ref`的引用对象改变为`b`，实际上并不是。它意味着：
```C++
a = b;
```
所以最后：
```
a = 20
b = 20
```
所以 **ref一直就是a的别名**，不能重新绑定

#### 总结：指针和引用的区别：
|         | Pointer        | Reference |
| ------- | -------------- | --------- |
| 保存什么    | 地址             | 对象别名      |
| 可为 null | 可以             | 正常情况下不应该  |
| 必须初始化   | 不必须            | 必须        |
| 可以重新指向  | 可以             | 不可以       |
| 使用对象    | `*ptr`         | `ref`     |
| 取地址     | `ptr` 本身就是地址   | `&ref`    |
| 常用于     | 可选对象、数组、底层 API | 参数、对象借用   |
